```python
from langchain_community.document_loaders import WebBaseLoader

url = "https://python.langchain.com/docs/introduction/"

loader = WebBaseLoader(web_paths=[url])

docs = loader.load()

print("Number of documents:", len(docs))
print("Content length:", len(docs[0].page_content))
print(docs[0].page_content[:10000])
```

Text Splitters

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
splits = text_splitter.split_documents(docs)
```

```python
print(splits[0])
print(splits[1])
print(splits[2])
```

```python
print(len(splits))
```

```python
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
import os

# Ensure sentence-transformers is installed for HuggingFaceEmbeddings
!pip install -qU sentence-transformers

# NOTE: groq_api_key is for the Groq Chat Model, not for embeddings.
# We will use HuggingFaceEmbeddings as an alternative to OpenAIEmbeddings.

# Initialize HuggingFaceEmbeddings. You can specify a different model if needed.
# A common model for general purpose embeddings is 'sentence-transformers/all-MiniLM-L6-v2'
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

vectorstore = Chroma.from_documents(documents=splits, embedding=embedding_model)
```

### Using Groq as an alternative to OpenAI

If you want to use Groq instead of OpenAI, you'll need to get a Groq API key. You can get one from [GroqCloud](https://console.groq.com/keys).

Once you have your key, store it in Colab secrets under the name `GROQ_API_KEY`.

```python
from google.colab import userdata
import os

# Get Groq API key from Colab secrets
groq_api_key = userdata.get('ChatGroq')

# Set it as an environment variable
os.environ['GROQ_API_KEY'] = groq_api_key

if not groq_api_key:
    raise ValueError("GROQ_API_KEY environment variable not set. Please set it in Colab secrets.")

print("Groq API key loaded successfully!")
```

Now you can initialize the `ChatGroq` model using your API key. Note that `ChatGroq` is a chat model, not an embedding model like `OpenAIEmbeddings`. If you need embeddings, you would typically use a separate embedding model (e.g., from Hugging Face) in conjunction with Groq for the chat model.

```python
from langchain_groq import ChatGroq

# Initialize the Groq chat model
llm = ChatGroq(
    model_name="llama3-8b-8192", # You can choose other models like 'mixtral-8x7b-32768'
    groq_api_key=groq_api_key
)

print(f"Groq Chat Model initialized with model: {llm.model_name}")
```
